## Introduction to Basic Machine Learning Workflow

### What We'll Cover
In the following cells, we will learn the essential steps of a machine learning workflow:

* Data Preprocessing (Scaling)
* Train-Test Split (Orniginal random split)
* Model Creation & Training
* Model Evaluation

### Step-by-Step Guide
1. Data Preprocessing    

    We'll first determine the features (X) and target (y) for our machine learning model.   
    Then we'll use StandardScaler to normalize our features. This helps models perform better by putting all features on similar scales.    

2. Train-Test Split    

    Splitting data into training (70%) and testing (30%) sets
    * Training data: Used to teach the model
    * Testing data: Used to evaluate model performance
3. Model Creation    

    Starting with Linear Regression as our baseline model. Simple but powerful for understanding relationships between variables.    
    Later, you'll implement:
    * Decision Trees (DT)
    * Random Forests (RF)
4. Model Evaluation
We'll use common metrics:
    * R-squared (R²) score
    * Mean Squared Error (MSE)
    * Mean Absorute Error (MAE)
    * Root Mean Squared Error (RMSE)

Let's begin with scaling our data in the next cell!

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error, root_mean_squared_error
```

```python
X_col = ["Wall_Rvalue","WWR","SHGC"] # Location should not be included in the feature
y_col = ["Total_Cooling_kwh"]
```

```python
## Data Preprocessing: Feature Scaling
X = df[X_col].copy()
y = result_df[y_col].copy()

scaler_wall_Rvalue = StandardScaler()
X["Wall_Rvalue"] = scaler_wall_Rvalue.fit_transform(df[["Wall_Rvalue"]])

scaler_WWR = StandardScaler()
X["WWR"] = scaler_WWR.fit_transform(df[["WWR"]])

scaler_SHGC = StandardScaler()
X["SHGC"] = scaler_SHGC.fit_transform(df[["SHGC"]])

scaler_y = StandardScaler()
y["Total_Cooling_kwh"] = scaler_y.fit_transform(result_df[["Total_Cooling_kwh"]])
```

```python
## train test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
## define metrics

metrics = ["MSE","R2","MAE","RMSE"]
```

Let's start with Linear Regression

```python
from sklearn.linear_model import LinearRegression
```

```python
# Initialize and train the linear regression model
model_linear = LinearRegression()
model_linear.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model_linear.predict(X_test)

# Inverse transform the scaled predictions and actual values back to original scale
y_pred_raw = scaler_y.inverse_transform(y_pred.reshape(-1, 1)).flatten()
y_test_raw = scaler_y.inverse_transform(y_test.values.reshape(-1, 1)).flatten()

# Calculate evaluation metrics
mse = mean_squared_error(y_test_raw, y_pred_raw)
mae = mean_absolute_error(y_test_raw, y_pred_raw)
r2 = r2_score(y_test_raw, y_pred_raw)
rmse = root_mean_squared_error(y_test_raw, y_pred_raw)

# Print the evaluation results
print("------ Linear prediction result ------")
print("Mean squared error: %.2f" % mse)
print("Mean absolute error: %.2f" % mae)
print("R2: %.2f" % r2)
print("Root Mean squared error: %.2f" % rmse)
```

```python
# define a dataframe to store results
results_df = pd.DataFrame(index=["Linear_Regression"],columns=metrics)
results_df.loc["Linear_Regression","MSE"] = mse
results_df.loc["Linear_Regression","MAE"] = mae
results_df.loc["Linear_Regression","R2"] = r2
results_df.loc["Linear_Regression","RMSE"] = rmse
results_df
```

### Add HDD and CDD and train new model

**Calculating Heating Degree Days (HDD) and Cooling Degree Days (CDD)**


HDD and CDD are metrics used to estimate energy demand for heating and cooling. They quantify how much the daily average temperature deviates from a base temperature (usually 65°F or 18°C).

**Calculation Method**

> Basic Formula    
> - **Daily HDD** = max(0, Base Temp - Daily Mean Temp)    
> - **Daily CDD** = max(0, Daily Mean Temp - Base Temp)    
> 
> Where:
> - **Daily Mean Temp** = (Max Temp + Min Temp) ÷ 2
> - **Base Temp** = Typically 65°F (18°C) - the temperature where heating/cooling becomes necessary
> 
> Example Calculation:    
> 
> If daily temperatures are: Max = 75°F, Min = 55°F
> - Mean Temp = (75 + 55) ÷ 2 = 65°F
> - HDD = max(0, 65 - 65) = 0
> - CDD = max(0, 65 - 65) = 0

```python
# calculate degree days
def calc_degree_days(temp_series, base_temp=18):
    hdd = (base_temp - temp_series).clip(lower=0)
    cdd = (temp_series - base_temp).clip(lower=0)
    return pd.DataFrame({'HDD': hdd, 'CDD': cdd})

CDD_dic = dict.fromkeys(Location)
HDD_dic = dict.fromkeys(Location)
for loc in Location:
    degree_day = calc_degree_days(weather_horly_df[loc], base_temp=18)
    weather_horly_df[loc+"_HDD"] = degree_day["HDD"]
    weather_horly_df[loc+"_CDD"] = degree_day["CDD"]
    CDD_dic[loc] = degree_day["CDD"].sum()
    HDD_dic[loc] = degree_day["HDD"].sum()
```

```python
# weather feature mapping for each record
weather_features_all = ["AVG_Temp","MAX_Temp","HDD","CDD"]
for feature in weather_features_all:
    df[feature] = df["Location"].map(eval(f"{feature}_dic"))
df.head()
### New model training
X_col_all = X_col + weather_features_all
print(X_col_all)
## Data Preprocessing: Feature Scaling
X_all = df[X_col_all].copy()

X_all["Wall_Rvalue"] = scaler_wall_Rvalue.transform(df[["Wall_Rvalue"]])
X_all["WWR"] = scaler_WWR.transform(df[["WWR"]])
X_all["SHGC"] = scaler_SHGC.transform(df[["SHGC"]])
X_all["AVG_Temp"] = scaler_AVG_Temp.transform(df[["AVG_Temp"]])
X_all["MAX_Temp"] = scaler_MAX_Temp.transform(df[["MAX_Temp"]])
scaler_CDD = StandardScaler()
X_all["CDD"] = scaler_CDD.fit_transform(df[["CDD"]])
scaler_HDD = StandardScaler()
X_all["HDD"] = scaler_HDD.fit_transform(df[["HDD"]])

# train test split (fixed random state to make sure the split is the same as before)
X_train_all, X_test_all, y_train_all, y_test_all = train_test_split(X_all, y, test_size=0.3, random_state=42)

## (optional) test if the split is the same as before
print("Is X_train the same as X_train_all? ", X_train.equals(X_train_all[X_col]))
print("Is X_test the same as X_test_all? ", X_test.equals(X_test_all[X_col]))
print("Is y_train the same as y_train_all? ", y_train.equals(y_train_all))
print("Is y_test the same as y_test_all? ", y_test.equals(y_test_all))
```

### Hyperparameter Tuning for Decision Trees and Random Forests

**Why Tune Hyperparameters?**
- Improve model performance and generalization
- Prevent overfitting/underfitting
- Optimize bias-variance tradeoff
- Enhance model interpretability

**Key Hyperparameters:**

**Decision Trees:**
- `max_depth`: Maximum depth of the tree
- `min_samples_split`: Minimum samples required to split a node
- `min_samples_leaf`: Minimum samples required at a leaf node
- `criterion`: Splitting criterion (gini, entropy)

**Random Forests:**
- `n_estimators`: Number of trees in the forest
- `max_features`: Number of features to consider for best split
- `bootstrap`: Whether to use bootstrap samples
- `max_samples`: Sample size for bootstrap if enabled

**Common Tuning Methods:**
- Grid Search (GridSearchCV)
- Random Search (RandomizedSearchCV)
- Bayesian Optimization
